```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import kmeans_plusplus, KMeans
```

```python
n_samples = 5_000
n_components = 5
```

```python
X,y_true = make_blobs(n_samples=n_samples,centers=n_components,cluster_std = 1.2,random_state=1)
X = X[:,::-1]
```

```python
centers_init, indices = kmeans_plusplus(X,n_clusters=5,random_state=1)
```

```python
plt.figure()
# colors = ["red","orange","m","#AA55B7","#11FF99","c","#986432","g"]
colors = ["red","orange","m","#AA55B7","#11FF99"]
for k,col in enumerate(colors):
    cluster_data = y_true == k
    plt.scatter(X[cluster_data,0],X[cluster_data,1],c=col,marker=".",s=10)

plt.scatter(centers_init[:,0],centers_init[:,1],c="b",s=60)
plt.title("algorytm k-means")
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
km = KMeans(n_clusters=5, init=centers_init,n_init=1,random_state=1)
km.fit(X)

y_pred = km.labels_
centers= km.cluster_centers_
```

```python
plt.figure(figsize=(8,6))

colors = ["red","orange","m","#AA55B7",
          "#11FF99"]

for k, col in enumerate(colors):
    cluster_data = (y_pred == k)
    plt.scatter(
        X[cluster_data, 0],
        X[cluster_data, 1],
        c=col,
        marker=".",
        s=10
    )

# centroidy
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    c="black",
    marker="X",
    s=200,
    label="Centroidy"
)

plt.title("Wynik algorytmu k-means")
plt.xticks([])
plt.yticks([])
plt.legend()
plt.show()
```
